```python
p_malware = 0.02
deteksi = 0.95
false_positive = 0.05

p_tidak = 1 - p_malware

p_peringatan = (
    deteksi * p_malware
    + false_positive * p_tidak
)

hasil = (deteksi * p_malware) / p_peringatan

print("Probabilitas peringatan:", p_peringatan)
print("Probabilitas malware:", hasil)
print("Persentase:", hasil * 100, "%")
```

```python
import numpy as np

jumlah_percobaan = 10000

hasil = np.random.choice(
    [1, 0],
    size=jumlah_percobaan,
    p=[0.8, 0.2]
)

berhasil = np.sum(hasil)
gagal = jumlah_percobaan - berhasil

probabilitas = berhasil / jumlah_percobaan

print("Jumlah percobaan:", jumlah_percobaan)
print("Berhasil:", berhasil)
print("Gagal:", gagal)
print("Probabilitas simulasi:", probabilitas)
```

```python
import scipy.stats as stats
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.1
x = 2

prob = stats.binom.pmf(x, n, p)
mean = stats.binom.mean(n, p)
variance = stats.binom.var(n, p)

print("Probabilitas:", prob)
print("Nilai harapan:", mean)
print("Varians:", variance)

nilai_x = np.arange(0, n + 1)
nilai_y = stats.binom.pmf(nilai_x, n, p)

plt.bar(nilai_x, nilai_y)
plt.title("Distribusi Binomial")
plt.xlabel("Jumlah Kegagalan")
plt.ylabel("Probabilitas")
plt.grid(axis="y")
plt.show()
```

```python
lamb = 5
x = 8

prob = stats.poisson.pmf(x, lamb)
mean = stats.poisson.mean(lamb)
variance = stats.poisson.var(lamb)

print("Probabilitas:", prob)
print("Nilai harapan:", mean)
print("Varians:", variance)

nilai_x = np.arange(0, 16)
nilai_y = stats.poisson.pmf(nilai_x, lamb)

plt.bar(nilai_x, nilai_y)
plt.title("Distribusi Poisson")
plt.xlabel("Jumlah Request")
plt.ylabel("Probabilitas")
plt.grid(axis="y")
plt.show()
```

```python
p = 0.2
x = 3

prob = stats.geom.pmf(x, p)
mean = stats.geom.mean(p)
variance = stats.geom.var(p)

print("Probabilitas:", prob)
print("Nilai harapan:", mean)
print("Varians:", variance)

nilai_x = np.arange(1, 16)
nilai_y = stats.geom.pmf(nilai_x, p)

plt.bar(nilai_x, nilai_y)
plt.title("Distribusi Geometrik")
plt.xlabel("Percobaan")
plt.ylabel("Probabilitas")
plt.grid(axis="y")
plt.show()
```

```python
mean = 100
std = 15

bawah = 85
atas = 115

prob = (
    stats.norm.cdf(atas, mean, std)
    - stats.norm.cdf(bawah, mean, std)
)

variance = stats.norm.var(mean, std)

print("Probabilitas:", prob)
print("Persentase:", prob * 100, "%")
print("Nilai harapan:", mean)
print("Varians:", variance)

x = np.linspace(40, 160, 500)
y = stats.norm.pdf(x, mean, std)

plt.plot(x, y)
plt.fill_between(
    x, y,
    where=(x >= bawah) & (x <= atas),
    alpha=0.4
)

plt.title("Distribusi Normal Waktu Respons")
plt.xlabel("Waktu Respons (ms)")
plt.ylabel("Kepadatan Probabilitas")
plt.grid()
plt.show()
```

```python
mean = 5
waktu = 10

prob = stats.expon.sf(waktu, scale=mean)

nilai_harapan = stats.expon.mean(scale=mean)
variance = stats.expon.var(scale=mean)

print("Probabilitas:", prob)
print("Persentase:", prob * 100, "%")
print("Nilai harapan:", nilai_harapan)
print("Varians:", variance)

x = np.linspace(0, 25, 500)
y = stats.expon.pdf(x, scale=mean)

plt.plot(x, y)
plt.fill_between(
    x, y,
    where=x >= waktu,
    alpha=0.4
)

plt.title("Distribusi Eksponensial")
plt.xlabel("Waktu Tunggu (Menit)")
plt.ylabel("Kepadatan Probabilitas")
plt.grid()
plt.show()
```

```python
a = 20
b = 100

bawah = 40
atas = 70

prob = (
    stats.uniform.cdf(atas, loc=a, scale=b-a)
    - stats.uniform.cdf(bawah, loc=a, scale=b-a)
)

mean = stats.uniform.mean(loc=a, scale=b-a)
variance = stats.uniform.var(loc=a, scale=b-a)

print("Probabilitas:", prob)
print("Persentase:", prob * 100, "%")
print("Nilai harapan:", mean)
print("Varians:", variance)

x = np.linspace(0, 120, 500)
y = stats.uniform.pdf(x, loc=a, scale=b-a)

plt.plot(x, y)
plt.fill_between(
    x, y,
    where=(x >= bawah) & (x <= atas),
    alpha=0.4
)

plt.title("Distribusi Uniform Kecepatan Download")
plt.xlabel("Kecepatan Download (Mbps)")
plt.ylabel("Kepadatan Probabilitas")
plt.grid()
plt.show()
```
